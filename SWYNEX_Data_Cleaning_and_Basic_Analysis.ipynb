```python
import pandas as pd
import numpy as np
import gzip
file_path = "/content/GSE2034_series_matrix.txt.gz"
import os

if os.path.exists(file_path):
    print("Dataset file found successfully!")
else:
    print("Dataset file not found. Please check the filename.")
```

```python
with gzip.open(file_path, "rt", encoding="utf-8",
               errors="replace") as f:
    lines = f.readlines()

print("File opened successfully!")
print("Total number of lines:", len(lines))
```

```python
start = next(
    i for i, line in enumerate(lines)
    if line.strip() == "!series_matrix_table_begin"
)

print("Data table starts at line:", start)
```

```python
end = next(
    i for i, line in enumerate(lines)
    if line.strip() == "!series_matrix_table_end"
)

print("Data table ends at line:", end)
```

```python
df = pd.read_csv(
    file_path,
    sep="\t",
    compression="gzip",
    skiprows=start + 1,
    nrows=end - start - 2
)

print("Dataset loaded successfully!")
```

```python
df.columns = [
    str(col).strip().strip('"')
    for col in df.columns
]

print(df.head())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Total probes:", df.shape[0])
print("Total samples:", df.shape[1] - 1)
```

```python
missing_values = df.isnull().sum().sum()

print("Total missing values:", missing_values)
```

```python
duplicate_rows = df.duplicated().sum()

print("Total duplicate rows:", duplicate_rows)
```

```python
duplicate_ids = df["ID_REF"].duplicated().sum()

print("Duplicate probe IDs:", duplicate_ids)
```

```python
# Select only the expression columns
numeric_data = df.iloc[:, 1:].apply(
    pd.to_numeric,
    errors="coerce"
)

# Check non-numeric or invalid values
print("Invalid values:", numeric_data.isnull().sum().sum())

# Check zero values
print("Zero values:", (numeric_data == 0).sum().sum())

# Check negative values
print("Negative values:", (numeric_data < 0).sum().sum())
```

```python
cleaned_df = df.copy()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)
print("Data cleaning completed successfully!")
```

```python
# Select gene expression measurements
expression_data = cleaned_df.iloc[:, 1:]

# Convert values into numbers
expression_data = expression_data.apply(
    pd.to_numeric,
    errors="coerce"
)

print("Numerical data prepared successfully!")
print("Shape:", expression_data.shape)
```

```python
mean_value = expression_data.mean().mean()

print("Mean expression value:", mean_value)
```

```python
median_value = expression_data.median().median()

print("Median expression value:", median_value)
```

```python
std_value = expression_data.stack().std()

print("Standard deviation:", std_value)
```

```python
minimum_value = expression_data.min().min()
maximum_value = expression_data.max().max()

print("Minimum value:", minimum_value)
print("Maximum value:", maximum_value)
```

```python
print("BASIC STATISTICAL ANALYSIS")
print("--------------------------")

print("Mean:", mean_value)
print("Median:", median_value)
print("Standard Deviation:", std_value)
print("Minimum:", minimum_value)
print("Maximum:", maximum_value)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))

values = expression_data.values.flatten()

plt.hist(values, bins=50)

plt.title("Gene Expression Distribution")
plt.xlabel("Expression Values")
plt.ylabel("Frequency")

plt.show()
```

```python
sample_means = expression_data.mean()

plt.figure(figsize=(10,5))

plt.plot(sample_means.values)

plt.title("Mean Gene Expression Across Samples")
plt.xlabel("Sample Number")
plt.ylabel("Mean Expression")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sample_medians = expression_data.median()

plt.boxplot(sample_medians)

plt.title("Distribution of Sample Median Expression")
plt.ylabel("Median Expression")

plt.show()
```

```python
# Save the cleaned dataset as CSV

cleaned_df.to_csv(
    "GSE2034_cleaned_dataset.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
# Create a statistical summary

statistics = pd.DataFrame({
    "Statistic": [
        "Mean",
        "Median",
        "Standard Deviation",
        "Minimum",
        "Maximum"
    ],
    "Value": [
        mean_value,
        median_value,
        std_value,
        minimum_value,
        maximum_value
    ]
})

statistics.to_csv(
    "statistical_summary.csv",
    index=False
)

print("Statistical summary saved!")
```
